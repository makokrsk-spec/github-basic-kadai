```python
class Human:
    def __init__(self, name, age):
        # 属性を定義する
        self.name = name
        self.age = age

    # メソッドprintinfoを使用する
    def printinfo(self):
        print(f"名前：{self.name}")
        print(f"年齢：{self.age}")

# 変数を代入する
human == Human(
    name="侍太郎",
    age=60
)

human.printinfo()
```
